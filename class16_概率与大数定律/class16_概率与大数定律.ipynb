
# 概率与大数定律

> **核心思想：世界存在不确定性，但不确定性可以通过概率、统计和数学模型进行描述。**

## 一、引言：从确定性到不确定性

### 1. 确定性与不确定性

在经典的确定性模型中，给定相同的输入和条件，通常会得到相同的输出。

例如：

- 给定函数 $f(x)=2x+1$，输入 $x=3$，输出必然是 7。
- 给定相同初始条件和理想模型，可以计算物体的运动轨迹。

但现实世界中还存在大量不确定现象：

- 掷骰子的结果无法事先准确预测；
- 网络数据包可能丢失；
- 保险客户是否发生事故存在不确定性；
- 机器学习模型会根据概率分布生成预测结果。

**概率论的任务之一，就是用数学语言描述和分析不确定性。**

### 2. 自然科学与人文社会科学

- **自然科学**：研究自然界的结构、现象及其规律，例如物理学、化学、生物学。
- **人文社会科学**：研究人类、社会、文化及其行为，例如历史学、社会学、经济学。

两者都需要逻辑推理、证据和方法论。技术发展也经常需要结合社会、伦理和人文视角，而不是简单地认为某一类学科必然取代另一类学科。

## 二、概率论的发展与相关人物

### 1. 帕斯卡与费马：赌徒问题

17 世纪，布莱士·帕斯卡（Blaise Pascal）与皮埃尔·德·费马（Pierre de Fermat）围绕“赌徒问题”展开讨论。

**赌徒问题：** 一场公平的比赛因故中断，如何根据双方当前的比赛进度，公平地分配赌注？

关键思想：

- 分析剩余比赛中各种可能的结果；
- 计算双方最终获胜的概率；
- 根据各自的获胜机会分配赌金。

这类问题推动了概率论的发展，也促进了期望思想的形成。

### 2. 雅各布·伯努利：大数定律

雅各布·伯努利（Jacob Bernoulli）在著作《推测术》（*Ars Conjectandi*）中系统讨论了大数定律。

核心思想：

> 在满足相应条件时，随着重复试验次数增加，事件的相对频率会趋近于它的概率。

例如，反复掷一枚公平硬币，正面出现的频率通常会随着试验次数增加而趋近于 \(1/2\)。

注意：频率不保证每次都单调接近概率，也不意味着有限次数下就一定非常接近。

### 3. 拉普拉斯：概率论的发展

皮埃尔-西蒙·拉普拉斯（Pierre-Simon Laplace）推动了概率论的发展与应用。

在**古典概率模型**中，如果样本空间包含有限个等可能结果，则事件 \(A\) 的概率为：

$$P(A)=\frac{|A|}{|S|}$$

其中：

- \(S\)：样本空间；
- \(A\)：目标事件；
- \(|A|\)：事件 \(A\) 包含的结果数量；
- \(|S|\)：样本空间中的结果总数。

需要注意：只有在有限且各结果等可能时，才能直接使用这个公式。

## 三、概率论在现实中的应用

### 1. 赌博与骰子

掷骰子时，单次结果无法准确预测，但可以描述各种结果出现的概率。

例如，对于一枚公平的六面骰子：

$$P(X=6)=\frac{1}{6}$$

概率衡量的是结果出现的可能性，而不是对单次结果的保证。

### 2. 保险与金融风险

保险公司需要估计事故发生概率、赔付金额及总体赔付风险，从而制定保费和准备金。

- 期望：衡量长期平均结果；
- 方差：衡量结果围绕期望的波动程度；
- 风险模型：结合概率、损失规模及其他业务因素进行评估。

高期望不一定意味着低风险。实际决策还要考虑损失分布、极端风险和风险承受能力。

### 3. 计算机科学中的随机性

- **网络**：分析数据包丢失、延迟和故障概率。
- **密码学**：需要安全的随机数或伪随机数生成机制；可预测的密钥会削弱安全性。
- **人工智能**：语言模型通常根据下一个 token 的概率分布进行采样。
- **随机化算法**：例如随机快速排序，通过随机选择主元，降低输入数据导致极端性能的风险。
- **蒙特卡洛方法**：通过随机采样估计复杂问题的数值结果。

### 4. 常见排序算法

排序算法不全是随机算法。常见排序算法包括：

## 四、概率的基本概念

### 1. 样本空间与事件

- **样本空间 \(S\)**：随机试验所有可能结果构成的集合。
- **事件 \(A\)**：样本空间的一个子集。
- **概率 \(P(A)\)**：事件 \(A\) 发生的可能性大小。

例如，掷一次六面骰子：

$$S=\{1,2,3,4,5,6\}$$

令 \(A\) 表示“掷出的点数为偶数”，则：

$$A=\{2,4,6\},\qquad P(A)=\frac{3}{6}=\frac{1}{2}$$

### 2. 概率的基本性质

$$0\le P(A)\le 1$$

$$P(S)=1,\qquad P(\varnothing)=0$$

如果事件 \(A\) 与 \(B\) 互斥，即 $(A\cap B=\varnothing)$，则：

$$P(A\cup B)=P(A)+P(B)$$

更一般地，两个事件的加法公式为：

$$P(A\cup B)=P(A)+P(B)-P(A\cap B)$$

### 3. 条件概率

在事件 \(B\) 已经发生且 \(P(B)>0\) 的条件下，事件 \(A\) 发生的概率为：

$$P(A\mid B)=\frac{P(A\cap B)}{P(B)}$$

条件概率改变了我们所考虑的信息范围。

### 4. 独立性

若事件 \(A\) 和 \(B\) 相互独立，则：

$$P(A\cap B)=P(A)P(B)$$

等价地，在 \(P(B)>0\) 时，独立意味着：

$$P(A\mid B)=P(A)$$

**互斥不等于独立。** 对概率都大于零的两个互斥事件，它们通常不独立，因为一个事件发生会排除另一个事件发生。

## 五、随机变量、期望与方差

### 1. 随机变量

随机变量 $(X)$ 是将随机试验结果映射为数值的函数。

例如，掷骰子时，令$(X)$ 表示骰子朝上的点数，则 $X\in\{1,2,3,4,5,6\}$。

### 2. 期望（Expectation）

期望是随机变量的概率加权平均值，可以理解为长期重复试验下的理论平均水平。

离散型随机变量：

$$E(X)=\sum_{i=1}^{n}x_iP(X=x_i)$$

连续型随机变量：

$$E(X)=\int_{-\infty}^{+\infty}xf_X(x)\,dx$$

其中 $f_X(x)$ 是随机变量的概率密度函数，积分存在时上述公式成立。

例如，公平六面骰子的期望为：

$$E(X)=\frac{1+2+3+4+5+6}{6}=3.5$$

注意：期望值不一定是随机变量实际可能取到的值。

### 3. 方差（Variance）

方差衡量随机变量围绕期望的平方偏差的平均水平。

$$\operatorname{Var}(X)=E[(X-E(X))^2]$$

等价公式：

$$\operatorname{Var}(X)=E(X^2)-[E(X)]^2$$

离散型随机变量：

$$\operatorname{Var}(X)=\sum_{i=1}^{n}(x_i-\mu)^2P(X=x_i)$$

其中 $\mu=E(X)$。

方差越大，说明取值相对于均值的波动通常越大。方差的单位是原变量单位的平方。

### 4. 期望与风险

在金融中，期望可以描述平均收益，方差可以描述波动程度，但不能只凭这两个指标就完整衡量风险。

- 高期望、高方差：平均收益可能较高，但波动也较大。
- 低期望、低方差：结果可能更稳定，但平均收益也可能较低。

极端损失、偏度、尾部风险和资金约束等也可能非常重要。

## 六、大数定律：连接理论概率与实际频率

### 1. 基本思想

设 $X_1,X_2,\ldots,X_n$ 为独立同分布随机变量，且期望存在，记：

$$E(X_i)=\mu$$

样本均值为：

$$\bar X_n=\frac{1}{n}\sum_{i=1}^{n}X_i$$

大数定律说明，在适当条件下，当样本量 \(n\) 增大时，样本均值会趋近总体期望：

$$\bar X_n\longrightarrow \mu$$

其中箭头的具体含义取决于所讨论的是弱大数定律还是强大数定律。

### 2. 伯努利试验中的大数定律

若每次试验成功的概率为 \(p\)，定义：

$$X_i=\begin{cases}1,&\text{第 }i\text{ 次试验成功}\\0,&\text{第 }i\text{ 次试验失败}\end{cases}$$

前 \(n\) 次试验的成功频率为：

$$\hat p_n=\frac{1}{n}\sum_{i=1}^{n}X_i$$

当试验满足独立同分布等条件时：

$$\hat p_n\longrightarrow p$$

也就是说，随着试验次数增加，成功频率趋近于成功概率。

### 3. 弱大数定律与强大数定律

**弱大数定律（Weak Law of Large Numbers）**

$$\forall\varepsilon>0,\quad \lim_{n\to\infty}P(|\bar X_n-\mu|>\varepsilon)=0$$

含义：当样本量趋向无穷时，样本均值偏离总体期望超过任意固定误差$\varepsilon$ 的概率趋于零。这叫作**依概率收敛**。

**强大数定律（Strong Law of Large Numbers）**

$$P\left(\lim_{n\to\infty}\bar X_n=\mu\right)=1$$

含义：样本均值以概率为 1 收敛到总体期望。这叫作**几乎必然收敛**。

在独立同分布且期望绝对可积等常见条件下，强大数定律成立。

注意：弱大数定律不是“趋近于 0”，而是偏离总体期望的概率趋近于 0；强大数定律中的“概率为 1”也不等同于有限步骤就必然精确等于均值。


## 七、切比雪夫不等式：偏离概率的上界

> **核心思想：随机变量偏离均值越远，偏离概率就越小；偏离概率的上界由方差决定。**

### 1. 基本公式

对于任意具有有限期望和有限方差的随机变量 \(X\)，以及任意 \(b>0\)，都有：

$$P(|X-E(X)|\geq b)\leq\frac{\operatorname{Var}(X)}{b^2}$$

其中：

- $(X)$：随机变量
- $(E(X)=\mu)$：随机变量的数学期望，即均值
- $(\operatorname{Var}(X)=\sigma^2)$：随机变量的方差
- $(b)$：允许偏离均值的距离
- $(P(|X-\mu|\geq b))$：随机变量与均值相差至少 \(b\) 的概率

**直观理解：** 方差越小，数据通常越集中；要求偏离均值的距离越大，偏离概率的上界越小。

### 2. 用标准差表示

由于标准差满足 $(\sigma=\sqrt{\operatorname{Var}(X)})$，当偏离距离为标准差的 $(k)$ 倍，即 $(b=k\sigma)$，且 $(k>0)$ 时：

$$P(|X-\mu|\geq k\sigma)\leq\frac{1}{k^2}$$

取补集，可得：

$$P(\mu-k\sigma<X<\mu+k\sigma)\geq 1-\frac{1}{k^2}$$

这意味着：**至少有 $(1-\frac{1}{k^2})$ 的概率，随机变量落在均值左右 \(k\) 个标准差的范围内。**

### 3. 典型结论

| 偏离均值的距离 | 偏离概率的上界 | 落在范围内的概率下界 |
|---|---:|---:|
| $(2\sigma)$ 及以上 | $(1/4=25\%)$ | 至少 $(75\%)$ |
| $(3\sigma)$ 及以上 | $(1/9\approx11.11\%)$ | 至少 $(88.89\%)$ |
| $(4\sigma)$ 及以上 | $(1/16=6.25\%)$ | 至少 $(93.75\%)$ |
| $(k\sigma)$ 及以上 | $(1/k^2)$ | 至少$(1-1/k^2)$ |

注意：这是**概率上界和下界**，不是说实际偏离比例一定达到这些数值。

### 4. 实际案例：考试成绩

假设一个班有 36 名学生，某次考试的平均分为 80 分，标准差为 10 分。

- 均值：$(\mu=80)$
- 标准差：$(\sigma=10)$
- 取 $(k=3)$，则偏离距离为 $(3\sigma=30)$ 分。

因此，偏离均值至少 30 分的成绩满足：

$$P(|X-80|\geq30)\leq\frac{1}{9}$$

也就是：

$$P(X\leq50\text{ 或 }X\geq110)\leq\frac{1}{9}$$

如果将每名学生的成绩视为一个观测值，那么偏离均值至少 30 分的学生人数不超过：

$$36\times\frac{1}{9}=4$$

所以，根据切比雪夫不等式，**至多有 4 名学生的成绩不高于 50 分或不低于 110 分**。

这里的 4 人是数学上界，不代表实际一定有 4 人；如果分数存在边界限制，例如满分 100 分，那么高于 110 分的情况自然不可能发生。

### 5. 与大数定律的关系

设 $(X_1,X_2,\ldots,X_n)$是相互独立、同分布的随机变量，且每个变量的期望为 $(\mu)$、方差为 $(\sigma^2)$。

样本均值为：

$$\overline{X}_n=\frac{1}{n}\sum_{i=1}^{n}X_i$$

样本均值的期望和方差分别为：

$$E(\overline{X}_n)=\mu,\qquad\operatorname{Var}(\overline{X}_n)=\frac{\sigma^2}{n}$$

将其代入切比雪夫不等式：

$$P(|\overline{X}_n-\mu|\geq\varepsilon)\leq\frac{\sigma^2}{n\varepsilon^2}$$

其中 $(\varepsilon>0)$ 是允许的误差。

当样本量 \(n\) 不断增大时，右侧趋近于 0：

$$\lim_{n\to\infty}P(|\overline{X}_n-\mu|\geq\varepsilon)=0$$

这说明样本均值偏离总体均值超过固定误差的概率会趋近于 0，是**切比雪夫不等式证明弱大数定律的经典思路**。

### 6. 核心总结

- **切比雪夫不等式：** 给出随机变量偏离均值的概率上界。
- **标准差倍数：** 偏离至少 $(k\sigma)$ 的概率不超过 $(1/k^2)$。
- **大数定律：** 样本量越大，样本均值偏离总体均值的概率越小。
- **重要区别：** 切比雪夫不等式提供的是保守的概率界限，并不要求**随机变量服从正态分布。**

## 八、中心极限定理：为什么正态分布很常见？

若 $(X_1,\ldots,X_n)$ 独立同分布，且具有有限均值 $(\mu)$ 和有限、非零方差 $(\sigma^2)$，则在经典中心极限定理条件下：

$$\frac{\sum_{i=1}^{n}X_i-n\mu}{\sigma\sqrt n}\xrightarrow{d}N(0,1)$$

等价地，对足够大的 \(n\)，样本均值常可近似为：

$$\bar X_n\approx N\left(\mu,\frac{\sigma^2}{n}\right)$$

其中 $(\xrightarrow{d})$ 表示依分布收敛，$(N(0,1)$ 表示标准正态分布。

直观理解：许多独立随机因素相加后，在满足适当条件时，标准化后的和会近似服从正态分布。

但不是所有数据本身都服从正态分布；中心极限定理讨论的是标准化后的和或均值的渐近分布。
## 十、蒙特卡洛模拟与大数定律

蒙特卡洛方法通过随机采样近似计算期望、概率或复杂数值。

例如，要估计事件 \(A\) 的概率，可以重复模拟 \(n\) 次，记录事件是否发生：

$$\hat P_n(A)=\frac{1}{n}\sum_{i=1}^{n}\mathbf{1}(A_i)$$

其中 \(\mathbf{1}(A_i)\) 是指示变量：事件发生时取 1，否则取 0。

在样本独立、分布正确且满足大数定律条件时：

$$\hat P_n(A)\longrightarrow P(A)$$

因此，蒙特卡洛估计会随着样本量增大而趋于稳定。不过，实际误差大小和收敛速度仍需分析。

应用示例：

1. **A/B 测试**：比较实验组与对照组的转化率等指标，并使用统计推断判断差异。
2. **围棋 AI**：蒙特卡洛树搜索会对候选行动进行模拟评估；AlphaGo 使用了蒙特卡洛树搜索，但还结合了深度神经网络等技术。
3. **大语言模型**：模型输出 logits，经温度等参数调整和概率归一化后，可按 token 概率分布采样。
4. **金融风控**：估计违约概率、损失分布和压力情景下的风险。

## 关键问题
### 一、大数定律与中心极限定理的区别

| 比较维度 | 大数定律                   | 中心极限定理                 |
| -------- | -------------------------- | ---------------------------- |
| 核心问题 | 样本均值是否趋近总体期望？ | 样本均值的误差如何分布？     |
| 主要结论 | \(\bar X_n\to\mu\)         | 标准化后的和趋近标准正态分布 |
| 关注重点 | 长期稳定性                 | 误差分布与近似计算           |
| 常见应用 | 频率估计、蒙特卡洛一致性   | 置信区间、统计推断、误差估计 |

一句话记忆：

- **大数定律：样本量足够大时，平均值会稳定到哪里？**
- **中心极限定理：平均值围绕目标波动的形状是什么？**

### 二、赌场为什么可能长期盈利？

赌场通常通过规则设计获得**庄家优势（House Edge）**。例如，在考虑全部规则后，赌场每下注 1 元的理论期望净收益可能为正。

设玩家每局净收益为 \(X_i\)，若各局独立同分布且期望为：

$$E(X_i)=-a,\qquad a>0$$

那么玩家的平均每局收益满足：

$$\frac{1}{n}\sum_{i=1}^{n}X_i\longrightarrow -a$$

这表示长期平均每局收益趋向负值，赌场对应的平均优势趋向正值。

关键不是“大数定律让赌场凭空赚钱”，而是：

1. 游戏规则使玩家的理论期望收益为负；
2. 大数定律使长期平均结果趋近理论期望；
3. 实际经营仍会受到波动、客流、成本和风险等因素影响。

**大数定律本身不保证赌场在任意有限时间内盈利，也不意味着每一局都会输。**

### 三、大数定律需要多少次试验才有效？

没有一个适用于所有问题的固定次数。所需样本量取决于：

- 目标误差 \(\varepsilon\)；
- 允许的失败概率；
- 方差大小；
- 分布形态和是否满足独立性等假设。

由切比雪夫不等式：

$$P(|\bar X_n-\mu|\ge\varepsilon)\le\frac{\sigma^2}{n\varepsilon^2}$$

如果希望偏差不超过 \(\varepsilon\) 的概率至少为 \(1-\delta\)，一个由该不等式给出的充分条件是：

$$n\ge\frac{\sigma^2}{\delta\varepsilon^2}$$

这是较保守的上界，不代表实际一定需要这么多样本。若已知更具体的分布或采用更精确的方法，通常可以得到更紧的估计。

### 四、概率为 0 的事件一定不会发生吗？

要区分离散型与连续型概率模型。

#### 1. 离散型情况

若事件由一个或多个离散结果构成，概率为 0 通常表示这些结果不可能发生，前提是模型已完整列出所有可能结果并正确赋予概率。

#### 2. 连续型情况

对于连续型随机变量，单个精确取值的概率可以为 0，但该取值仍属于可能出现的结果集合。

例如，若 \(X\sim U(0,1)\)，则：

$$P(X=0.5)=0$$

但 \(X\) 仍然可以取到区间 \((0,1)\) 内的值，包括 \(0.5\)。连续分布为区间赋予概率，而不是为单个点赋予正概率。

因此，**概率为 0 不总是等同于逻辑上不可能发生**。


### 五、Python：生成符合概率分布的数据

使用 NumPy 可以方便地进行随机采样。

#### 1. 正态分布：`np.random.normal`

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

# 生成 10,000 个样本：均值为 0，标准差为 1
samples = rng.normal(loc=0, scale=1, size=10_000)

print("样本均值：", samples.mean())
print("样本方差：", samples.var())

plt.hist(samples, bins=50, density=True)
plt.title("Samples from the standard normal distribution")
plt.xlabel("x")
plt.ylabel("Density")
plt.show()
```

参数含义：

- `loc`：均值；
- `scale`：标准差，而不是方差；
- `size`：样本数量。

#### 2. 二项分布：`np.random.binomial`

```python
import numpy as np

rng = np.random.default_rng(42)

# 进行 100 次独立试验，每次成功概率为 0.3
# 重复进行 10,000 组试验
samples = rng.binomial(n=100, p=0.3, size=10_000)

print("理论均值：", 100 * 0.3)
print("样本均值：", samples.mean())
print("理论方差：", 100 * 0.3 * (1 - 0.3))
print("样本方差：", samples.var())
```

二项分布的理论均值和方差分别为：

$$E(X)=np,\qquad \operatorname{Var}(X)=np(1-p)$$

#### 3. 逆变换采样

如果 $(U\sim U(0,1))$，且目标分布的累积分布函数为 $(F)$，那么在适当条件下，可以通过：

$$X=F^{-1}(U)$$

生成服从目标分布的随机变量。

例如，指数分布的累积分布函数为 $(F(x)=1-e^{-\lambda x})$，其中 $(x\ge0)$、$(\lambda>0)$。令 $(U\sim U(0,1))$，可得到：

$$X=-\frac{\ln(1-U)}{\lambda}$$

代码如下：

```python
import numpy as np

rng = np.random.default_rng(42)

lam = 2.0
u = rng.random(10_000)

# 逆变换采样：生成指数分布样本
samples = -np.log1p(-u) / lam

print("样本均值：", samples.mean())
print("理论均值：", 1 / lam)
```

`np.log1p(-u)` 用于更稳定地计算 $(\ln(1-u))$。
