- **1683 年：关孝和**
  - 日本数学家

  - 著作《解伏题之法》中出现了行列式的雏形

- **莱布尼茨**
  - 德国数学家
  - 研究线性方程组和消元法
  - 发现方程组解中的分母可以由系数按照一定规律组合得到
  - 逐渐形成了行列式的思想

## 二、凯莱与矩阵代数

### 1. 凯莱

凯莱系统地整理和发展了**矩阵代数**，包括：

- 矩阵乘法
- 单位矩阵
- 逆矩阵
- 矩阵的代数运算
- 矩阵与行列式之间的关系

### 2. 逆矩阵

对于方阵 $A$，如果存在矩阵 $A^{-1}$，满足：

$$
\boxed{
AA^{-1}=A^{-1}A=I
}
$$

那么 $A^{-1}$ 称为 $A$ 的**逆矩阵**。

其中：

- $A^{-1}$：逆矩阵
- $I$：单位矩阵

逆矩阵**存在**的条件：

$$
\boxed{
\det(A)\neq0
}
$$

因此：

$$
\boxed{
\det(A)\neq0
\Longleftrightarrow
A\text{ 可逆}
}
$$

而：

$$
\boxed{
\det(A)=0
\Longleftrightarrow
A\text{ 不可逆}
}
$$

## 三、伴随矩阵

### 1. adj(A)

$$
\operatorname{adj}(A)
$$

称为 $A$ 的**伴随矩阵(adjugate matrix)**：


### 2. 伴随矩阵如何得到？

#### 第一步：求每个元素的代数余子式

$$
C_{ij}=(-1)^{i+j}M_{ij}
$$

其中：

- $M_{ij}$：余子式
- $C_{ij}$：代数余子式

#### 第二步：用代数余子式组成矩阵

$$
C=
\begin{pmatrix}
C_{11}&C_{12}&\cdots\\
C_{21}&C_{22}&\cdots\\
\vdots&\vdots&\ddots
\end{pmatrix}
$$

#### 第三步：转置

$$
\boxed{
\operatorname{adj}(A)=C^T
}
$$

所以：

> **伴随矩阵 = 代数余子式矩阵的转置**

---

## 四、利用伴随矩阵求逆

当：

$$
\det(A)\neq0
$$

时：

$$\boxed{A^{-1}=\frac{1}{\det(A)}\operatorname{adj}(A)}$$

其中：

- $\det(A)$：矩阵的行列式
- $\operatorname{adj}(A)$：伴随矩阵
- $A^{-1}$：逆矩阵

核心关系：

$$\boxed{A\operatorname{adj}(A)=\operatorname{adj}(A)A=\det(A)I}$$

因此：

$$
A
\left(
\frac{\operatorname{adj}(A)}{\det(A)}
\right)
=I
$$

所以：

$$\boxed{A^{-1}=\frac{\operatorname{adj}(A)}{\det(A)}}$$

### 一句话理解

> **行列式决定能不能求逆，伴随矩阵负责构造逆矩阵。**

## 五、$2\times2$ 矩阵求逆

设：

$$
A=
\begin{pmatrix}
a&b\\
c&d
\end{pmatrix}
$$

### 行列式

$$
\boxed{
\det(A)=ad-bc
}
$$

### 伴随矩阵

$$\boxed{\operatorname{adj}(A)=\begin{pmatrix}d&-b\\-c&a\end{pmatrix}}$$

### 逆矩阵

$$\boxed{A^{-1}=
\frac{1}{ad-bc}
\begin{pmatrix}
d&-b\\
-c&a
\end{pmatrix}
}
$$

前提：

$$
\boxed{
ad-bc\neq0
}
$$

## 六、方阵才有普通意义上的行列式

普通行列式：

$$\det(A)$$

定义在**方阵**上。

例如：

$$
A_{2\times2}
$$

可以计算：

$$
\det(A)
$$

但是：

$$
A_{2\times3}
$$

不是方阵，因此没有通常意义上的：

$$
\det(A)
$$

所以：

> **普通行列式属于方阵。**

# 七、行列式的几何意义

## 1. 二维：面积缩放

一个边长为 $1$ 的正方形：

$$
1\times1
$$

经过线性变换：

$$
\mathbf{x}'=A\mathbf{x}
$$

会变成一个平行四边形。

二阶行列式表示二维线性变换对**面积的缩放比例**：

$$\boxed{|\det(A)|=\text{面积缩放倍数}}$$

例如：

$$
\det(A)=3
$$

表示面积扩大为原来的 $3$ 倍。

---

## 2. 三维：体积缩放

对于 $3\times3$ 矩阵：

$$\boxed{|\det(A)|=\text{体积缩放倍数}}$$

所以：

- $\det(A)>1$：体积扩大
- $0<\det(A)<1$：体积缩小
- $\det(A)<0$：体积缩放的同时发生方向翻转
- $\det(A)=0$：空间被压缩到更低维

例如三维立方体经过线性变换后：

$$
\text{立方体}
\rightarrow
\text{平行六面体}
$$

如果：

$$
\det(A)=0
$$

则可能变成：

$$
\text{三维空间}
\rightarrow
\text{二维平面}
$$

甚至：

$$
\text{三维空间}
\rightarrow
\text{一条直线}
$$

因此信息发生丢失，矩阵不可逆。

---

# 八、$\det(A)\neq0$ 到底意味着什么？

$$
\boxed{
\det(A)\neq0
}
$$

意味着：

1. $A$ **可逆**
2. $A^{-1}$ 存在
3. $A$ 满秩
4. $A$ 的行向量线性无关
5. $A$ 的列向量线性无关
6. 线性变换没有发生降维
7. 方程组 $Ax=b$ **有唯一解**

所以可以记成：

$$
\boxed{
\det(A)\neq0
\Longleftrightarrow
A^{-1}\text{ 存在}
}
$$

而：

$$
\boxed{
\det(A)=0
\Longleftrightarrow
A\text{ 不可逆}
}
$$

---

# 九、行列式的常用性质

## 1. 转置

$$
\boxed{
\det(A^T)=\det(A)
}
$$

矩阵转置后：

$$
\text{行}\leftrightarrow\text{列}
$$

但行列式的值不变。

---

## 2. 矩阵乘法

$$
\boxed{
\det(AB)=\det(A)\det(B)
}
$$

几何上：

> 组合变换的面积/体积缩放倍数 = 各个变换缩放倍数的乘积。

---

## 3. 逆矩阵

如果 $A$ 可逆：

$$
AA^{-1}=I
$$

两边取行列式：

$$
\det(A)\det(A^{-1})=\det(I)
$$

因为：

$$
\det(I)=1
$$

所以：

$$\boxed{\det(A^{-1})=\frac{1}{\det(A)}}$$

---

# 十、行变换与行列式

### 1. 交换两行

$$
\boxed{
\det(A')=-\det(A)
}
$$

交换两行，行列式**变号**。

---

### 2. 某一行乘以 $k$

$$
\boxed{
\det(A')=k\det(A)
}
$$

---

### 3. 某一行加上另一行的 $k$ 倍

$$
\boxed{
\det(A')=\det(A)
}
$$

---

### 记忆

> **换行变号，倍乘跟着变，加倍数不变。**

---

# 十一、行列式与线性方程组

对于方阵线性方程组：

$$
Ax=b
$$

如果：

$$
\det(A)\neq0
$$

那么：

$$
\boxed{
x=A^{-1}b
}
$$

并且方程组有唯一解。

如果：

$$
\det(A)=0
$$

则 $A$ 不可逆，方程组可能：

- 无解
- 有无穷多个解

因此：

> **行列式可以判断方阵线性方程组是否具有唯一解。**

---

# 十二、克莱姆法则

对于：

$$
Ax=b
$$

如果：

$$
\det(A)\neq0
$$

则：

$$
\boxed{
x_i=
\frac{\det(A_i)}{\det(A)}
}
$$

其中：

- $A_i$：把 $A$ 的第 $i$ 列替换成 $b$ 得到的矩阵

### 特点

克莱姆法则：

- 数学形式非常优美
- 适合理论证明
- 手算低维方程组可以使用
- 高维情况下计算量很大
- 工程计算中一般不会使用克莱姆法则直接求解大型方程组

工程上通常使用：

- LU
- QR
- SVD 等数值方法

---

# 十三、高斯-若尔当法求逆矩阵

构造增广矩阵：

$$
[A\mid I]
$$

通过初等行变换，把左边的 $A$ 变成单位矩阵：

$$
\boxed{
[A\mid I]
\quad\Longrightarrow\quad
[I\mid A^{-1}]
}
$$

例如：

$$
\left[
\begin{array}{cc|cc}
a&b&1&0\\
c&d&0&1
\end{array}
\right]
\Longrightarrow
\left[
\begin{array}{cc|cc}
1&0&*&*\\
0&1&*&*
\end{array}
\right]
$$

右边最终得到：

$$
A^{-1}
$$

### 本质

> **用高斯消元法，同时求解多个线性方程组，从而得到逆矩阵。**

复杂度通常为：

$$
\boxed{
O(n^3)
}
$$

相比直接使用伴随矩阵计算，更适合计算机数值计算。

---

# 十四、为什么工程上一般不用伴随矩阵求逆？

理论上：

$$A^{-1}=\frac{\operatorname{adj}(A)}{\det(A)}
$$

但是高维矩阵需要计算大量余子式和行列式，计算量非常大。

因此：

> **伴随矩阵求逆适合数学推导，不适合高维工程计算。**

工程中通常根据问题选择：

- LU 分解
- QR 分解
- SVD
- 直接求解线性方程组，而不是显式计算 $A^{-1}$

---

# 十五、LU 分解

$$
\boxed{
A=LU
}
$$

其中：

- $L$：下三角矩阵
- $U$：上三角矩阵

### 特点

- 主要用于方阵
- 计算速度较快
- 常用于求解线性方程组
- 实际计算通常使用**部分主元选取**

一般写成：

$$
\boxed{
PA=LU
}
$$

其中 $P$ 是置换矩阵。

### 为什么需要主元？

如果当前主元：

$$
a_{kk}=0
$$

就不能直接进行消元。

即使：

$$
a_{kk}\neq0
$$

但如果它非常接近 $0$，也可能带来较大的数值误差。

因此实际计算中通常通过交换行来选择合适的主元。

---

# 十六、QR 分解

$$
\boxed{
A=QR
}
$$

其中：

- $Q$：正交矩阵
- $R$：上三角矩阵

对于方阵：

$$
Q^TQ=I
$$

因此：

$$
\boxed{
Q^{-1}=Q^T
}
$$

### Q 的特点

- 每一列都是单位向量
- 不同列之间相互正交
- 任意两列：

$$
q_i^Tq_j=0
\quad(i\neq j)
$$

- 正交变换不会改变向量长度
- 数值稳定性较好

### R 的特点

上三角矩阵：

$$
R=
\begin{pmatrix}
r_{11}&r_{12}&r_{13}\\
0&r_{22}&r_{23}\\
0&0&r_{33}
\end{pmatrix}
$$

---

## QR 分解的作用

### 1. 解线性方程组

$$
Ax=b
$$

由于：

$$
A=QR
$$

所以：

$$
QRx=b
$$

左乘 $Q^T$：

$$
Rx=Q^Tb
$$

最后求解上三角方程组。

---

### 2. 最小二乘法

QR 分解最重要的应用之一：

$$
\boxed{
\text{最小二乘拟合}
}
$$

特别适合：

$$
Ax\approx b
$$

这种通常没有精确解的超定方程组。

---

### 3. QR 迭代

QR 算法是经典的特征值计算方法。

通过不断进行：

$$
A_k=Q_kR_k
$$

然后：

$$
A_{k+1}=R_kQ_k
$$

经过迭代，矩阵逐渐趋向于 **Schur 形式**。

对于实对称矩阵，可以进一步趋向于对角矩阵，其对角线元素就是特征值。

---

## QR 分解如何获得？

### Gram-Schmidt 正交化

- 原理简单
- 适合教学和理解

### Householder 反射

- 数值稳定
- 工业计算中常用

### Givens 旋转

- 每次只消去一个元素
- 适合稀疏矩阵和需要局部修改的情况

---

# 十七、SVD 奇异值分解

SVD：

> Singular Value Decomposition
> 奇异值分解

### 核心思想

任意矩阵都可以分解为：

$$
\boxed{
A=U\Sigma V^T
}
$$

对于一个 $m\times n$ 矩阵：

$$\boxed{A_{m\times n}=
U_{m\times m}
\Sigma_{m\times n}
V^T_{n\times n}
}
$$

其中：

- $A$：原矩阵
- $U$：左奇异向量矩阵
- $\Sigma$：奇异值矩阵
- $V^T$：右奇异向量矩阵的转置

---

## SVD 的几何理解

可以理解为：

$$\boxed{\text{旋转}\rightarrow\text{坐标轴缩放}\rightarrow\text{再旋转}}$$

即：

$$
A = U\Sigma V^T
$$

其中：

- $V^T$：第一次旋转
- $\Sigma$：沿不同方向进行缩放
- $U$：第二次旋转

---

## 奇异值矩阵

$$
\Sigma=
\begin{pmatrix}
\sigma_1&&&\\
&\sigma_2&&\\
&&\ddots&\\
&&&\sigma_r\\
&&&\\
\end{pmatrix}
$$

其中：

$$
\sigma_1\geq\sigma_2\geq\cdots\geq\sigma_r\geq0
$$

这些：

$$
\sigma_1,\sigma_2,\cdots,\sigma_r
$$

称为**奇异值**。

如果某个奇异值：

$$
\sigma_i=0
$$

说明对应方向的信息被完全压缩掉了。

---

# 十八、SVD 的应用

### 1. 降维与 PCA

PCA 与 SVD 密切相关。

可以通过保留最大的前 $k$ 个奇异值：

$$
\sigma_1,\sigma_2,\cdots,\sigma_k
$$

进行低秩近似：

$$
\boxed{
A\approx A_k
}
$$

从而实现：

- 降维
- 数据压缩
- 图像压缩
- 噪声过滤

---

### 2. Moore-Penrose 伪逆

对于非方阵：

$$
A_{m\times n}
$$

通常不存在普通意义上的：

$$
A^{-1}
$$

但是可以通过 SVD 构造：

$$
\boxed{
A^+
}
$$

即 Moore-Penrose 伪逆。

可以用于：

- 超定方程组
- 欠定方程组
- 最小二乘问题

---

### 3. 条件数

对于适当的矩阵：

$$\boxed{\kappa(A)=\frac{\sigma_{\max}}{\sigma_{\min}}}$$

奇异值可以用来判断矩阵的数值稳定性。

如果：

$$
\sigma_{\min}\approx0
$$

那么：

$$
\kappa(A)
$$

可能非常大，说明矩阵可能存在严重的数值病态问题。

---

### 4. 推荐系统与 NLP

SVD 的低秩近似思想可以用于：

- 推荐系统
- 潜在语义分析 LSA
- NLP
- 矩阵降维
- 噪声过滤

---

# 十九、LU、QR、SVD 对比

| 分解 | 主要适用 | 核心特点 | 计算成本 |
|---|---|---|---|
| **LU** | 主要是方阵 | 快速求解线性方程组 | 较低 |
| **QR** | 方阵、长方阵 | 数值稳定，适合最小二乘 | 高于 LU |
| **SVD** | 任意矩阵 | 最全面，可处理秩亏和病态问题 | 最高 |

### 一句话记忆

$$
\boxed{
LU\rightarrow 快
}
$$

$$
\boxed{
QR\rightarrow 稳
}
$$

$$
\boxed{
SVD\rightarrow 全
}
$$

# 二十一、计算中的行列式

## 1. 判断矩阵是否可逆

$$
\boxed{
\det(A)\neq0
\Rightarrow
A^{-1}\text{ 存在}
}
$$

$$
\boxed{
\det(A)=0
\Rightarrow
A^{-1}\text{ 不存在}
}
$$

---

## 2. 数值稳定性的预警

如果：

$$
\det(A)=0
$$

说明矩阵奇异。

如果某些情况下矩阵接近奇异：

$$
\det(A)\approx0
$$

可能意味着数值计算存在风险。

但要注意：

> **判断矩阵是否病态，不能只看 $\det(A)$ 是否接近 0，更常用条件数或最小奇异值。**

---

# 二十二、已知变换结果，求变换前的输入

假设：

$$
y=Ax
$$

现在已经知道 $y$，想要求 $x$：

$$
Ax=y
$$

如果：

$$
A^{-1}\text{ 存在}
$$

则：

$$
\boxed{
x=A^{-1}y
}
$$

可以把逆矩阵理解成：

> **把原来的线性变换反过来。**

例如：

- 图像变换
- 3D 图形变换
- 坐标变换
- 密码学

---

# 二十三、应用

### 1. 计算机图形学

3D 图形中的：

- 平移
- 旋转
- 缩放
- 坐标变换

都大量使用矩阵。

---

### 2. 密码学

例如希尔密码：

$$
C=KP
$$

其中：

- $K$：密钥矩阵
- $P$：明文
- $C$：密文

解密：

$$
\boxed{
P=K^{-1}C
}
$$

因此密钥矩阵必须满足：

$$
\boxed{
\det(K)\neq0
}
$$

才能存在逆矩阵。

---

### 3. 机器学习

矩阵和行列式可以出现在：

- 线性回归
- 高斯混合模型
- 协方差矩阵
- 优化
- 最小二乘

---

### 4. 机器人与控制系统

机器人运动和控制中大量使用矩阵。

例如机械臂的雅可比矩阵：

$$
J
$$

可以描述关节速度与末端执行器速度之间的关系：

$$
\dot{x}=J\dot{q}
$$

当雅可比矩阵出现秩亏时，机器人可能进入**奇异位形**，某些运动方向的控制能力会丢失。

```python
A = np.array([[2.0, 1.0], [1.0, 3.0]])
# 避免int 带来的计算异常
print('det 手算 2*3-2*1=', 2 * 3 - 2 * 1)
print('det numpy = ', np.linalg.det(A))
# 求矩阵的逆矩阵
Ainv = np.linalg.inv(A)
print("A的逆矩阵 \n", Ainv)
# @ 矩阵的乘法  浮点数精度误差导致的 导致负数的0
print('A @ 逆 约等于\n', np.round(A @ Ainv, 10))
```

```python
# 奇异矩阵(Singular matrix)
B = np.array([[1.0, 2.0], [2.0, 4.0]])
print("det(B)", np.linalg.det(B))

try:
    print(np.linalg.solve(B, np.array([1.0, 2.0])))

except np.linalg.LinAlgError as e:
    print('solve  Error:', e)

try:
    print('inv(B)', np.linalg.inv(B))
except np.linalg.LinAlgError as e:
    print('inv  Error:', e)

# 只要det(X) 等于0 ,solve,inv都会报 Singular matrix 错误
```

```python
# 面积几何


v1, v2 = np.array([2.0, 0.5]), np.array([0.5, 2.0])
M = np.column_stack([v1, v2])
# 行列式的几何意义有两个列向量围成几何的面积
det = np.linalg.det(M)
print('列向量矩阵 det = ', det, '有向面积')
plt.figure(figsize=(5, 5))
pts = np.array([[0, 0], v1, v1 + v2, v2, [0, 0]])

plt.fill(pts[:, 0], pts[:, 1], alpha=0.3)
plt.plot([0, v1[0]], [0, v1[1]], '-o', label='v1')
plt.plot([0, v2[0]], [0, v2[1]], '-o', label='v2')
# 坐标轴设置等比例的
plt.gca().set_aspect('equal')

plt.grid(True, alpha=0.3)
plt.legend()
plt.title(f'平行四边形面积 |det|={abs(det):.4f}')
plt.show()

```

```python
# 可视化推荐书籍

A = np.array([[2.0, 1.0], [1.0, 3.0]])
b = np.array([5.0, 5.0])

# 底层使用LU分解,浮点误差小,更快
x1 = np.linalg.solve(A, b)
# 求逆计算量太大,且精度也会下降
x2 = np.linalg.inv(A) @ b
print('solve:', x1)
print('inv@b:', x2)
print('相同?', np.allclose(x1, x2))
print('优先solve')
```
