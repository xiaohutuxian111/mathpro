```python
"""
 逻辑运算运用
"""

# 程序判断就是逻辑运算
age = 20
has_id =  True

#  p 交q ：年龄满足且有证件的才能进入

if  age >=18 and  has_id:
    print("可以进入")
else:
    print("不可以进入")

```

```python
# 德摩根定律在代码的运用

# 非（p交q） = 非p 并 非q

if not(age >=18 and has_id):
    print("被拒绝")
if age< 18 or not has_id:
    print("被拒绝")
```
