# 特征值与特征向量

> **矩阵的 DNA：研究矩阵内部最底层的特征。**

前面：

- **逆矩阵**：可以恢复线性变换
- **行列式**：告诉我们线性变换对空间整体的缩放，以及是否发生空间坍缩
- **特征值与特征向量**：进一步研究矩阵变换中的特殊方向以及这些方向上的缩放倍数

---

## 1. 特征值与特征向量的直观理解

矩阵进行线性变换时，一般情况下：

> 向量的方向和长度都会发生改变。

但存在一些特殊方向，经过矩阵变换后：

> **方向不变，只改变长度。**

可以把这样的向量理解成一根“钉子”，固定在原来的直线上。

这个特殊向量就是：

**特征向量（Eigenvector）**

而长度改变的倍数就是：

**特征值（Eigenvalue）**

因此：

> **特征向量代表矩阵变换中的特殊方向，特征值代表该方向上的缩放倍数。**

---

## 2. 数学定义

设矩阵 $A$ 存在一个非零向量 $\mathbf v$，使得：

$$
\boxed{A\mathbf v=\lambda\mathbf v}
$$

其中：

- $A$：矩阵
- $\mathbf v$：特征向量
- $\lambda$：特征值
- $\mathbf v\neq0$

含义：

> 矩阵 $A$ 作用在特征向量 $\mathbf v$ 上，相当于沿着 $\mathbf v$ 所在方向进行 $\lambda$ 倍缩放。

例如：

- $\lambda>1$：沿该方向拉长
- $0<\lambda<1$：沿该方向压缩
- $\lambda=1$：长度不变
- $\lambda<0$：方向反向，同时按 $|\lambda|$ 缩放
- $\lambda=0$：该方向被压缩到零向量

---

# 3. 特征方程

从特征值定义开始：

$$
A\mathbf v=\lambda\mathbf v
$$

移项：

$$
A\mathbf v-\lambda\mathbf v=0
$$

因为：

$$
\lambda\mathbf v=\lambda I\mathbf v
$$

所以：

$$
(A-\lambda I)\mathbf v=0
$$

由于特征向量要求：

$$
\mathbf v\neq0
$$

因此，$(A-\lambda I)$ 必须存在非零解。

也就是说：

$$
\det(A-\lambda I)=0
$$

因此得到：

$$
\boxed{\det(A-\lambda I)=0}
$$

这就是**特征方程**。

> 特征方程把**行列式**和**特征值**联系起来，特征值就是这个方程的根。

---

# 4. 特征值与特征向量的求解过程

整个过程可以记为：

$$
A\mathbf v=\lambda\mathbf v
$$

↓

$$
(A-\lambda I)\mathbf v=0
$$

↓

$$
\det(A-\lambda I)=0
$$

↓

求出特征值：

$$
\lambda
$$

↓

再将 $\lambda$ 代入：

$$
(A-\lambda I)\mathbf v=0
$$

↓

求出对应的特征向量：

$$
\mathbf v
$$

---

# 5. 例子：计算矩阵的特征值与特征向量

给定矩阵：

$$
A=
\begin{pmatrix}
2&1\\
1&2
\end{pmatrix}
$$

---

## 5.1 写出特征方程

根据：

$$
\det(A-\lambda I)=0
$$

得到：

$$
A-\lambda I=
\begin{pmatrix}
2-\lambda&1\\
1&2-\lambda
\end{pmatrix}
$$

因此：

$$
\begin{vmatrix}
2-\lambda&1\\
1&2-\lambda
\end{vmatrix}
=0
$$

---

## 5.2 计算行列式

对于 $2\times2$ 矩阵：

$$
\begin{vmatrix}
a&b\\
c&d
\end{vmatrix}
=ad-bc
$$

所以：

$$
(2-\lambda)(2-\lambda)-1\times1=0
$$

即：

$$
(2-\lambda)^2-1=0
$$

---

## 5.3 展开

展开：

$$
4-4\lambda+\lambda^2-1=0
$$

得到：

$$
\lambda^2-4\lambda+3=0
$$

---

## 5.4 因式分解

$$
(\lambda-1)(\lambda-3)=0
$$

因此得到两个特征值：

$$
\boxed{\lambda_1=1,\qquad\lambda_2=3}
$$

也就是说，这个矩阵有两个特征方向：

- $\lambda=3$：沿对应特征方向放大 $3$ 倍
- $\lambda=1$：沿对应特征方向长度保持不变

---

# 6. 求对应的特征向量

得到特征值之后，将特征值分别代入：

$$
(A-\lambda I)\mathbf v=0
$$

---

## 6.1 当 $\lambda=3$

代入：

$$
A-3I=
\begin{pmatrix}
-1&1\\
1&-1
\end{pmatrix}
$$

设：

$$
\mathbf v=
\begin{pmatrix}
x\\
y
\end{pmatrix}
$$

则：

$$
\begin{pmatrix}
-1&1\\
1&-1
\end{pmatrix}
\begin{pmatrix}
x\\
y
\end{pmatrix}
=
\begin{pmatrix}
0\\
0
\end{pmatrix}
$$

得到：

$$
-x+y=0
$$

因此：

$$
x=y
$$

可以取：

$$
\boxed{
\mathbf v_1=
\begin{pmatrix}
1\\
1
\end{pmatrix}
}
$$

验证：

$$
A
\begin{pmatrix}
1\\
1
\end{pmatrix}
=
\begin{pmatrix}
3\\
3
\end{pmatrix}
=
3
\begin{pmatrix}
1\\
1
\end{pmatrix}
$$

所以：

$$
\boxed{
\lambda=3,\qquad
\mathbf v=
\begin{pmatrix}
1\\
1
\end{pmatrix}
}
$$

---

## 6.2 当 $\lambda=1$

代入：

$$
A-I=
\begin{pmatrix}
1&1\\
1&1
\end{pmatrix}
$$

设：

$$
\mathbf v=
\begin{pmatrix}
x\\
y
\end{pmatrix}
$$

则：

$$
\begin{pmatrix}
1&1\\
1&1
\end{pmatrix}
\begin{pmatrix}
x\\
y
\end{pmatrix}
=
\begin{pmatrix}
0\\
0
\end{pmatrix}
$$

得到：

$$
x+y=0
$$

因此：

$$
x=-y
$$

可以取：

$$
\boxed{
\mathbf v_2=
\begin{pmatrix}
1\\
-1
\end{pmatrix}
}
$$

验证：

$$
A
\begin{pmatrix}
1\\
-1
\end{pmatrix}
=
\begin{pmatrix}
1\\
-1
\end{pmatrix}
=
1
\begin{pmatrix}
1\\
-1
\end{pmatrix}
$$

所以：

$$
\boxed{
\lambda=1,\qquad
\mathbf v=
\begin{pmatrix}
1\\
-1
\end{pmatrix}
}
$$

---

# 7. 这个例子的几何意义

对于：

$$
A=
\begin{pmatrix}
2&1\\
1&2
\end{pmatrix}
$$

我们得到：

$$
\lambda_1=3,\qquad
\mathbf v_1=
\begin{pmatrix}
1\\
1
\end{pmatrix}
$$

以及：

$$
\lambda_2=1,\qquad
\mathbf v_2=
\begin{pmatrix}
1\\
-1
\end{pmatrix}
$$

因此：

### 方向 $\begin{pmatrix}1\\1\end{pmatrix}$

沿着：

$$
y=x
$$

这个方向，矩阵将向量放大 **3 倍**。

### 方向 $\begin{pmatrix}1\\-1\end{pmatrix}$

沿着：

$$
y=-x
$$

这个方向，矩阵的作用相当于：

> **长度保持不变，方向也保持不变。**

因此，这两个方向就是矩阵的两个**特征方向**。

---

# 8. 特征向量：系统的“主轴”

特征向量可以理解为系统中的：

> **“主轴”或“固有方向”**

矩阵沿着这些方向进行变换时，不会把方向扭转成其他方向，而只会进行缩放。

因此：

$$
\boxed{\text{特征向量}=\text{固有方向}}
$$

$$
\boxed{\text{特征值}=\text{该方向上的缩放倍数}}
$$

---

# 9. 历史人物

特征值与特征向量的思想是在数学、力学等领域的发展过程中逐渐形成的。

## 9.1 欧拉

**欧拉（Euler）**

18世纪，瑞士数学家。

研究刚体旋转时发现：

> 刚体旋转存在特殊的旋转轴。

这个**主轴**可以看作后来特征向量“特殊方向”思想的早期直觉。

即：

> **主轴的方向 → 特殊方向 → 特征向量的思想**

---

## 9.2 拉格朗日

**拉格朗日（Lagrange）**

18世纪，法国数学家。

研究力学、天体运动等问题时，可以通过选择合适的**坐标系**，将复杂的运动问题分解成更加简单的独立问题。

这种思想与后来矩阵的：

> **对角化、解耦**

有密切联系。

---

## 9.3 柯西

**柯西（Cauchy）**

19世纪，法国数学家。

对行列式、矩阵理论以及特征方程的发展做出了重要贡献。

特征方程：

$$
\boxed{\det(A-\lambda I)=0}
$$

它建立了：

> **行列式 ↔ 特征值**

之间的联系。

### 对称矩阵

对于实对称矩阵：

$$
A=A^T
$$

其特征值为实数，并且不同特征值对应的特征向量相互正交。

---

## 9.4 希尔伯特

**希尔伯特（Hilbert）**

在积分方程和谱理论的发展中发挥了重要作用。

“Eigenvalue”来自德语：

$$
\text{Eigenwert}
$$

其中：

- **Eigen**：自己的、固有的
- **Wert**：值

因此可以理解为：

> **Eigenvalue：矩阵自身固有的值**

对应：

> **Eigenvector：矩阵自身固有的方向**

# 10. 固有频率的直觉

特征值与特征向量不仅存在于抽象数学中，也可以用来描述物理系统的**固有特性**。

例如：

### 吉他弦

吉他弦振动时存在特定的振动频率和振动模式。

可以类比为：

- 特征值 → 固有频率
- 特征向量 → 固有振动模式

类似的例子还有：

- 桥梁振动
- 弹簧振动
- 建筑抗震
- 固有频率
- 固有模态

这些系统都需要分析其特殊的振动模式。

# 11. 为什么需要特征值？

现实中的系统通常存在大量变量。

变量之间相互影响，就形成了：

## 耦合（Coupling）

两个或多个变量之间相互依赖、相互影响。

数学上表现为：

> 多个方程相互关联，不能直接拆开。

在计算机领域中，也可以用“耦合”描述模块、代码之间的依赖程度。

一般来说：

- 内部耦合
- 公共耦合
- 控制耦合
- 数据耦合

其中数据耦合相对较低，只通过简单参数传递数据，不直接依赖内部细节。

---

## 解耦（Decoupled）

特征值分解的一个重要思想：

> **寻找合适的特征方向，把复杂系统分解成相对独立的部分。**

因此：

$$
\boxed{
\text{复杂系统}
\rightarrow
\text{寻找特征方向}
\rightarrow
\text{解耦}
}
$$

特征向量代表系统的“主轴”。

## 12.6 PageRank

Google 的 PageRank 可以看作特征向量的一个典型应用。

将互联网看作一个图：

- 网页 → 节点
- 网页之间的链接 → 边

定义一个随机游走过程：

> 用户不断点击网页链接，在经过大量跳转后，用户停留在各个页面的概率逐渐趋于稳定。

这个稳定的概率向量可以表示为：

$$
A\mathbf p=\mathbf p
$$

也就是：

$$
A\mathbf p=1\mathbf p
$$

因此：

$$
\boxed{\lambda=1}
$$

对应的特征向量就是稳定的概率向量。

这也是 Google 搜索排名背后的重要数学思想之一。

# 13. 特征值、特征向量、特征方程三者关系

可以用一句话串起来：

$$
\boxed{A\mathbf v=\lambda\mathbf v}
$$

是**特征值与特征向量的定义**。

为了找到 $\lambda$：

$$
\boxed{\det(A-\lambda I)=0}
$$

是**特征方程**。

找到 $\lambda$ 后，再求：

$$
\boxed{(A-\lambda I)\mathbf v=0}
$$

得到对应的**特征向量**。
